```python
#handle data types for date
def clean_date(df):
    #change dtype and infer format in the date column
    df['date'] = pd.to_datetime(df['date'],errors ='coerce',dayfirst = False, format = "%d/%m/%Y")
    return df
# use our function
df = clean_date(df)
```

```python
#handle datatype and string characters in the amount column
def convert_to_usd(df):
    # Ensure all values in the 'amount' column are strings
    df['amount'] = df['amount'].astype(str)
    
    # Define a function to convert currencies to USD
    def convert_currency(amount):
        if '€' in amount:
            return float(amount.strip('€')) / 0.94  # Euro to USD
        elif '$' in amount:
            return float(amount.strip('$'))  # USD is already in USD
        elif '£' in amount:
            return float(amount.strip('£')) / 0.82  # GBP to USD
        elif 'JPY' in amount:
            return float(amount.strip('JPY'))/ 135  # Japanese Yen to USD
        elif 'INR' in amount:
            return float(amount.strip('INR'))/ 83  # Indian Rupee to USD
        else:
            return float(amount)  # In case there is no currency symbol

    # Apply the conversion function to the 'amount' column
    df['amount'] = df['amount'].apply(convert_currency).round(2)
    # Rename the column
    df.rename(columns={'amount': 'amount_dollars'}, inplace=True)
    
    return df

# use our function
df = convert_to_usd(df)
```

```python
# handle the datatype and string characters in the temperature column
def convert_to_celcius(df):
    # Ensure all values in the 'temp' column are strings
    df['temperature'] = df['temperature'].astype(str)
    
    # Define a function to convert Fahrenheit to Celsius
    def convert_degrees(temp):
        if '°F' in temp:
            return (float(temp.replace('°F',''))-32)*5/9  # Fahrenheit to Celsius
        elif '°C' in temp:
            return float(temp.replace('°C','')) 
        else:
            return float(temp)  # In case there is no symbol

    # Apply the conversion function 
    df['temperature'] = df['temperature'].apply(convert_degrees).round(2)
    # Rename the column
    df.rename(columns={'temperature': 'temperature_celsius'}, inplace=True)
    
    return df

# use our function
df = convert_to_celcius(df)
```

```python
#handle value inconsistencies
def fix_categories(df):
    #remove white spaces and capitalise
    df['day']= df['day'].astype(str).str.strip().str.capitalize()
    replacements =  {'Wed':'Wednesday','Fri':'Friday','Tues':'Tuesday','Sat':'Saturday'}
    df['day'] = df['day'].replace(replacements)
    #flight status
    df['flight_status']= df['flight_status'].astype(str).str.strip().str.capitalize()
    df['flight_status'] = df['flight_status'].replace('On time','On_time')
    #gender
    df['gender']= df['gender'].astype(str).str.strip().str.capitalize()
    df['gender']= df['gender'].replace('Nb','Non-binary')
    #travel purpose
    df['travel_purpose']= df['travel_purpose'].astype(str).str.strip().str.capitalize()
    #checking method
    df['checkin_method'] = df['checkin_method'].astype(str).str.strip().str.capitalize()
    return df
# use our function
df = fix_categories(df)
```

```python
#read in dataset
airline_df = pd.read_csv('airline_df.csv')
#imputing missing values
airline_df.head(2)
```

```python
#handle missing values.since we have only 5% of observations we are to opt for basic imputation techniques
from sklearn.impute import SimpleImputer
def missing_values(airline_df):
    #replace 'Nan' strings as np.nan
    airline_df.replace('Nan',np.nan, inplace = True)
    #delete the date column for that half of its values are missing
    airline_df.drop(columns ='date', inplace = True )
    #fill missing values
    #categorical columns
    cat_columns = airline_df.select_dtypes('object').columns
    imputer = SimpleImputer(strategy='most_frequent')
    airline_df[cat_columns] = imputer.fit_transform(airline_df[cat_columns])
    #numerical
    num_columns = airline_df.select_dtypes( include =['float64','int64']).columns
    imputer = SimpleImputer(strategy='mean')
    airline_df[num_columns] = imputer.fit_transform(airline_df[num_columns])
    
    return airline_df
#use our function
airline_df = missing_values(airline_df)
```

<h3>feature engineering<h3>

```python
#map day column into weekday and weekend
def feature_manipulation(df):
    #mapping days to categories
    weekday_map = {
    'Monday': 'Weekday', 'Tuesday': 'Weekday', 'Wednesday': 'Weekday','Thursday': 'Weekday', 'Friday': 'Weekday', 
    'Saturday': 'Weekend', 'Sunday': 'Weekend'}
    airline_df['day_type'] = airline_df['day'].map(weekday_map)
    #notice the temp column has too many decimals
    airline_df['temperature_celsius']= airline_df['temperature_celsius'].round(3)
    #drop low prices in amount_dollars. in real life there is no plain ticket below 10 dollars
    low_amounts =  airline_df[ airline_df['amount_dollars']<=10]
    airline_df.drop(low_amounts.index, axis = 0, inplace = True)
       
    return df
#use our function
airline_df = feature_manipulation(airline_df)
airline_df.head()
```
